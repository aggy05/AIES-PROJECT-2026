```python
def extract_text_from_pdf(filepath):
    with pdfplumber.open(filepath) as pdf:
        text = ""
        for page in pdf.pages:
            t = page.extract_text()
            text += t if t else ""
    return text
```

```python
skills_database = {
    "python": ["python"],
    "machine learning": ["machine learning", "ml"],
    "deep learning": ["deep learning", "dl"],
    "natural language processing": ["natural language processing", "nlp"],
    "data analysis": ["data analysis", "data analytics"],
    "artificial intelligence": ["artificial intelligence", "ai"],
    "sql": ["sql", "mysql", "postgresql"],
    "java": ["java"],
    "javascript": ["javascript", "js"],
    "html": ["html"],
    "css": ["css"],
    "excel": ["excel", "ms excel"],
    "communication": ["communication"],
    "teamwork": ["teamwork", "team player"],
    "problem solving": ["problem solving"],
    "data science": ["data science"],
    "tensorflow": ["tensorflow"],
    "keras": ["keras"],
    "pandas": ["pandas"],
    "numpy": ["numpy"],
    "power bi": ["power bi"],
    "tableau": ["tableau"],
    "c++": ["c++"],
    "r": [" r ", "r programming"],
}

def extract_skills(text, skills_db):
    text_lower = text.lower()
    found = []
    for skill, synonyms in skills_db.items():
        for synonym in synonyms:
            if synonym in text_lower:
                found.append(skill)
                break
    return found

def match_resume_to_job(resume_skills, job_skills):
    matched = [s for s in resume_skills if s in job_skills]
    missing = [s for s in job_skills if s not in resume_skills]
    score = round((len(matched) / len(job_skills)) * 100, 2) if job_skills else 0
    return matched, missing, score
```

```python
def generate_ai_feedback(resume_text, matched_skills, missing_skills, score):
    feedback = []
    text_lower = resume_text.lower()

    # ── SCORE BASED FEEDBACK ──────────────────────────
    if score >= 70:
        feedback.append("✅ STRONG MATCH: Your resume aligns well with the job requirements. Focus on polishing presentation.")
    elif score >= 40:
        feedback.append("⚠️  MODERATE MATCH: Your resume partially matches. Targeted improvements will significantly boost your chances.")
    else:
        feedback.append("❌ LOW MATCH: Your resume needs significant work to align with this role. Follow the suggestions below carefully.")

    # ── RESUME LENGTH CHECK ───────────────────────────
    word_count = len(resume_text.split())
    if word_count < 200:
        feedback.append("📄 LENGTH: Your resume seems very short. A strong resume should have at least 400–600 words covering your skills, experience, and projects.")
    elif word_count > 1000:
        feedback.append("📄 LENGTH: Your resume is quite long. Try to keep it concise — ideally 1 page for freshers, 2 pages max for experienced candidates.")
    else:
        feedback.append("📄 LENGTH: Your resume length looks good.")

    # ── QUANTIFICATION CHECK ──────────────────────────
    numbers_present = any(char.isdigit() for char in resume_text)
    if not numbers_present:
        feedback.append("📊 IMPACT: Your resume lacks numbers and metrics. Add quantifiable achievements like 'Improved accuracy by 20%' or 'Managed a team of 5' to stand out.")
    else:
        feedback.append("📊 IMPACT: Good — your resume includes numbers/metrics which strengthens your impact statements.")

    # ── PROJECT EXPERIENCE CHECK ──────────────────────
    if "project" in text_lower:
        feedback.append("🔨 PROJECTS: Good — you have project experience mentioned. Make sure each project clearly states the problem, tools used, and outcome.")
    else:
        feedback.append("🔨 PROJECTS: No projects detected. Adding 2–3 relevant projects significantly improves your resume, especially if you have limited work experience.")

    # ── EDUCATION CHECK ───────────────────────────────
    if any(word in text_lower for word in ["bachelor", "master", "b.e", "b.tech", "mba", "degree", "university", "college"]):
        feedback.append("🎓 EDUCATION: Education section detected. Make sure your CGPA or grade is mentioned if it's above 7.0/3.5.")
    else:
        feedback.append("🎓 EDUCATION: Education section not clearly detected. Ensure your degree, university, and graduation year are clearly mentioned.")

    # ── CERTIFICATIONS CHECK ──────────────────────────
    if any(word in text_lower for word in ["certified", "certification", "certificate", "coursera", "udemy", "nptel"]):
        feedback.append("🏅 CERTIFICATIONS: Great — certifications found. These add credibility especially for technical roles.")
    else:
        feedback.append("🏅 CERTIFICATIONS: No certifications detected. Adding relevant certifications (Coursera, Udemy, NPTEL) can strengthen your profile.")

    # ── MISSING SKILLS SUGGESTIONS ────────────────────
    suggestions = {
        "python": "Learn Python at https://www.learnpython.org",
        "machine learning": "Take Andrew Ng's ML course on Coursera — highly recommended.",
        "deep learning": "Try the Deep Learning Specialization on Coursera or fast.ai.",
        "natural language processing": "Explore HuggingFace NLP course at https://huggingface.co/learn",
        "data analysis": "Practice with pandas and real datasets on Kaggle.com",
        "artificial intelligence": "Start with AI For Everyone by Andrew Ng on Coursera.",
        "sql": "Practice SQL free at https://sqlbolt.com",
        "communication": "Add group projects, presentations, or volunteer work to show communication skills.",
        "problem solving": "Practice on LeetCode or HackerRank and mention it on your resume.",
        "tensorflow": "Start with TensorFlow tutorials at https://www.tensorflow.org/tutorials",
        "keras": "Learn Keras at https://keras.io/getting_started",
        "pandas": "Practice pandas at https://pandas.pydata.org/docs",
        "numpy": "Learn NumPy at https://numpy.org/learn",
        "power bi": "Free Power BI learning at https://learn.microsoft.com/en-us/power-bi",
        "tableau": "Free Tableau training at https://www.tableau.com/learn/training",
        "data science": "Try IBM Data Science Professional Certificate on Coursera.",
        "excel": "Free Excel training at https://support.microsoft.com/en-us/excel",
        "java": "Learn Java at https://www.learnjavaonline.org",
        "javascript": "Start JavaScript at https://javascript.info",
        "html": "Learn HTML at https://www.w3schools.com/html",
        "css": "Learn CSS at https://www.w3schools.com/css",
    }

    if missing_skills:
        feedback.append("\n🎯 MISSING SKILLS — Here's how to fill the gaps:")
        for skill in missing_skills:
            tip = suggestions.get(skill, f"Consider learning {skill.title()} through online platforms like Coursera or Udemy.")
            feedback.append(f"   • {skill.upper()}: {tip}")

    return feedback
```

```python
# Change this to any resume file in /content
resume_path = "/content/Data_Science_and_Analytics_2.pdf"

job_description = """
We are hiring a Machine Learning Engineer.
The candidate must have strong Python and SQL skills.
Knowledge of TensorFlow, Keras, and deep learning is required.
Experience with data analysis and NLP is a plus.
Strong problem solving and communication skills are essential.
"""

# Process
resume_text = extract_text_from_pdf(resume_path)
resume_skills = extract_skills(resume_text, skills_database)
job_skills = extract_skills(job_description, skills_database)
matched, missing, score = match_resume_to_job(resume_skills, job_skills)

# Display Report
print("=" * 60)
print("        AI-POWERED RESUME FEEDBACK REPORT")
print("=" * 60)
print(f"\nMatch Score   : {score}%")
print(f"Matched Skills: {', '.join(matched) if matched else 'None'}")
print(f"Missing Skills: {', '.join(missing) if missing else 'None'}")
print("\n── AI FEEDBACK ──────────────────────────────────────")

feedback = generate_ai_feedback(resume_text, matched, missing, score)
for point in feedback:
    print(f"\n{point}")

print("\n" + "=" * 60)
```
