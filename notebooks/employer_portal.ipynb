```python
skills_database = {
    "python": ["python"],
    "machine learning": ["machine learning", "ml"],
    "deep learning": ["deep learning", "dl"],
    "natural language processing": ["natural language processing", "nlp"],
    "data analysis": ["data analysis", "data analytics"],
    "artificial intelligence": ["artificial intelligence", "ai"],
    "sql": ["sql", "mysql", "postgresql"],
    "java": ["java"],
    "javascript": ["javascript", "js"],
    "html": ["html"],
    "css": ["css"],
    "excel": ["excel", "ms excel"],
    "communication": ["communication"],
    "teamwork": ["teamwork", "team player"],
    "problem solving": ["problem solving"],
    "data science": ["data science"],
    "tensorflow": ["tensorflow"],
    "keras": ["keras"],
    "pandas": ["pandas"],
    "numpy": ["numpy"],
    "power bi": ["power bi"],
    "tableau": ["tableau"],
    "c++": ["c++"],
    "r": [" r ", "r programming"],
}

def extract_text_from_pdf(filepath):
    with pdfplumber.open(filepath) as pdf:
        text = ""
        for page in pdf.pages:
            t = page.extract_text()
            text += t if t else ""
    return text

def extract_skills(text, skills_db):
    text_lower = text.lower()
    found = []
    for skill, synonyms in skills_db.items():
        for synonym in synonyms:
            if synonym in text_lower:
                found.append(skill)
                break
    return found
```

```python
# Employer posts a job — paste any job description here
job_description = """
We are hiring a Machine Learning Engineer.
The candidate must have strong Python and SQL skills.
Knowledge of TensorFlow, Keras, and deep learning is required.
Experience with data analysis and NLP is a plus.
Strong problem solving and communication skills are essential.
"""

job_skills = extract_skills(job_description, skills_database)

print("Skills required for this job:")
for skill in job_skills:
    print(" •", skill)
print(f"\nTotal required: {len(job_skills)} skills")
```

```python
# Put all candidate resumes as PDFs inside a folder called "resumes"
resume_folder = "/content"

resume_files = [
    f for f in os.listdir(resume_folder) if f.endswith(".pdf")
]

print(f"Found {len(resume_files)} resumes to process:")
for f in resume_files:
    print(" •", f)
```

```python
def match_resume_to_job(resume_skills, job_skills):
    matched = [s for s in resume_skills if s in job_skills]
    missing = [s for s in job_skills if s not in resume_skills]
    score = round((len(matched) / len(job_skills)) * 100, 2) if job_skills else 0
    return matched, missing, score

results = []

for filename in resume_files:
    filepath = os.path.join(resume_folder, filename)
    text = extract_text_from_pdf(filepath)
    resume_skills = extract_skills(text, skills_database)
    matched, missing, score = match_resume_to_job(resume_skills, job_skills)

    results.append({
        "Candidate": filename.replace(".pdf", ""),
        "Match Score (%)": score,
        "Matched Skills": ", ".join(matched) if matched else "None",
        "Missing Skills": ", ".join(missing) if missing else "None",
        "Total Skills Found": len(resume_skills)
    })

print("All resumes processed!")
```

```python
# Sort candidates by match score (highest first)
df = pd.DataFrame(results)
df = df.sort_values("Match Score (%)", ascending=False).reset_index(drop=True)
df.index += 1  # Rank starts from 1

print("=" * 60)
print("        CANDIDATE RANKING — EMPLOYER VIEW")
print("=" * 60)

for rank, row in df.iterrows():
    print(f"\nRank #{rank}  —  {row['Candidate']}")
    print(f"  Match Score  : {row['Match Score (%)']}%")
    print(f"  Matched      : {row['Matched Skills']}")
    print(f"  Missing      : {row['Missing Skills']}")

print("\n")
df
```
